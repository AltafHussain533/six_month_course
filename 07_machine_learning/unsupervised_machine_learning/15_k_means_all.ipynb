# **K-means Clustering**

* K-mean clustering is a type of unsupervised learning algorithm.
* It works on unlabeled data. 
* It make clusters from unlabeled data.

## **Example**

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import seaborn as sns

#ignore warnings
import warnings
warnings.filterwarnings("ignore")

# generate sample data
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.6, random_state=0)

# initiate a KMeans object with the number of clusters
kmeans = KMeans(n_clusters=4)
# fit the data to the KMeans algorithm
kmeans.fit(X)

# get the predicted cluster labels for each data point
labels = kmeans.labels_

# plot the data points with different colors for different clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title('K-Means Clustering')
plt.show()
```

# Parallel coordinate plots
**Parallel coordinate plots** are a visualization tool used to analyze multivariate data, which means data with many different variables or columns. Instead of using perpendicular axes (like the X and Y axes on a standard graph), this plot places all axes parallel to each other, allowing you to compare dozens of variables at the same time.
## How Parallel Coordinate Plots Work
- **The Layout:**Each variable (column) gets its own vertical (or horizontal) axis. These axes sit side-by-side.
- **The Points:** A single row of data (a data point) is drawn as a continuous line that connects across all the axes.
- **The Interaction:** The point where the line crosses each axis represents the value for that specific variable
 ### Common Real-World Uses
Parallel coordinate plots are highly valuable for finding clusters, patterns, and outliers in complex datasets:
- **Data Science & Machine Learning:** Used to visualize high-dimensional data before training models, helping data scientists see which features correlate or group together.
- **Automotive & Product Design:** Engineers use them for multi-objective optimization. For example, comparing car models across price, fuel efficiency, horsepower, safety rating, and weight all at once to find the best trade-off.
-  **Finance & Portfolio Analysis:** Comparing stocks or investment funds across multiple metrics like risk, return, volatility, dividend yield, and market cap.
-  **Genomics & Bioinformatics:** Analyzing how thousands of genes express themselves across different experimental conditions or patient samples.
-  **Quality Control:** Tracking manufacturing data across multiple sensor checks simultaneously to pinpoint exactly where a product failed the assembly line.
### Key Advantages and Challenges
- **The Big Pro:** You are not limited to 2D or 3D graphs. You can easily plot 10, 20, or more variables on a single screen.
- **The Main Con:** They can easily look like a tangled "hairball" if you have thousands of rows of data.
To make them readable, data analysts often use interactive tools to filter specific lines, color-code the lines by category, or drag the axes around to see how adjacent relationships change

The **elbow method** is a visual technique used in data science to find the optimal number of clusters for K-Means clustering.\
It helps you balance model simplicity with accuracy by identifying the point where adding more clusters no longer gives significantly better results.\
**1. Calculate Clustering Inertia**\
You run the K-Means algorithm multiple times, each time increasing the number of clusters (K), typically from 1 to 10. For each run, you calculate the Sum of Squared Errors (SSE), also known as inertia. This metric measures how tightly packed the data points are around their respective cluster centers.\
As the number of clusters increases, the inertia will naturally decrease toward zero.

# **Model Evaluation** 
### 1. **Elbow Method**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Load the Diamonds dataset
df = sns.load_dataset('diamonds')

# Select relevant features for clustering
features = ['carat', 'depth', 'price']

# Extract the selected features from the dataset
X = df[features]

# Perform K-means clustering for different values of k
k_values = range(1, 11)  # Test k from 1 to 10
inertias = []  # List to store the inertia values
# inertia sum of squared within clusters

for k in k_values:
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)

# Plot the Elbow curve
plt.plot(k_values, inertias, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

# In elbow method why we see elbow point
We look for the elbow point because it represents the mathematical "sweet spot" where you get the maximum reward for the minimum amount of complexity.
It is the exact moment where the model stops capturing meaningful patterns and starts over-segmenting the data.
1. The Law of Diminishing Returns
Think of adding clusters like hiring workers to clean a messy room:
- Hiring worker 1 to 3 (Before the Elbow): The room gets clean incredibly fast. The error (inertia) drops drastically because big, obvious groups of data are being separated.
- Hiring worker 4 (The Elbow): The room is clean. This is the optimal number of people.
- Hiring workers 5 to 10 (After the Elbow): They have nothing major left to do. They start splitting small, clean piles into even smaller piles. The error barely drops, but now you are paying for 10 workers instead of 4.
- 2. Balancing Fit vs. Simplicity
In machine learning, you always fight a battle between two things:
1. Model Accuracy: You want your clusters to be tight and accurate.
2. Model Simplicity: You want as few clusters as possible so humans can actually interpret and use the data.\
If you have 100 data points and you choose (K=100) clusters, your error drops to exactly zero because every point gets its own cluster. However, a model with 100 clusters is completely useless for business or analysis. The elbow point shows you where you have successfully minimized error without making the model overly complex.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Load the Diamonds dataset
df = sns.load_dataset('diamonds')

# Select relevant features for clustering
features = ['carat', 'depth', 'price']

# Extract the selected features from the dataset
X = df[features]

# Perform K-means clustering for different values of k
k_values = range(1, 11)  # Test k from 1 to 10
inertias = []  # List to store the inertia values

for k in k_values:
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)

# Plot the Elbow curve
plt.plot(k_values, inertias, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')

# Find the elbow point
diff = [inertias[i] - inertias[i+1] for i in range(len(inertias)-1)]
best_k = diff.index(max(diff)) + 2

# Add a vertical line to indicate the elbow point
plt.axvline(x=best_k, color='r', linestyle='--', label='Elbow Point')

plt.legend()
plt.show()

print("The best number of clusters based on the Elbow Method is:", best_k)
```

### 2. **Silouette Method**

The **Silhouette Method** is a technique used in data science to find the optimal number of clusters for K-Means clustering. Unlike the elbow method, which only looks at how tight individual clusters are, the silhouette method measures how well-separated and distinct the clusters are from each other.
It calculates a Silhouette Coefficient for each data point, which ranges from -1 to +1.1. 
### 1. **How the Silhouette Score is Calculated** 
For any individual data point i, the method calculates two critical distances:\
  2. Intra-cluster distance **(ai)**: The average distance between point i and all other points in the same cluster. You want this number to be as small as possible (tight clusters).\
  3. Nearest-cluster distance **(bi)**: The average distance between point i and all points in the nearest neighboring cluster. You want this number to be as large as possible (well-separated clusters).
 The Silhouette Coefficient (si) for that single point is calculated using the formula:
 $$s_i = \frac{b_i - a_i}{\max(a_i, b_i)}$$

 The overall silhouette score for a specific number of clusters (K) is simply the average score of all data points combined.
### 1. **How to Interpret the Scores**
When you plot the average silhouette score for different values of K (e.g., K = 2 to 10), you look for the highest peak on the graph.
- Score near +1: The point is far away from neighboring clusters and very close to its own. This indicates excellent clustering.
- Score near 0: The point sits right on the boundary between two clusters. The clusters are overlapping.
- Score near -1: The point has been placed in the wrong cluster entirely, as it is closer to a neighboring cluster than its own.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Load the Diamonds dataset
df = sns.load_dataset('diamonds')

# Select relevant features for clustering
features = ['carat', 'depth', 'price']

# Extract the selected features from the dataset
X = df[features]

# Perform K-means clustering for different values of k
k_values = range(2, 11)  # Test k from 2 to 10
silhouette_scores = []  # List to store the silhouette scores

for k in k_values:
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X)
    labels = kmeans.labels_
    silhouette_scores.append(silhouette_score(X, labels))

# Plot the Silhouette scores
plt.plot(k_values, silhouette_scores, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Method')
plt.show()
# that way k_means is computically expensive and time consuming not run this code in our pc.
#  it may take a long time to run, especially for larger datasets.
```

# **K-means++**

The main difference between K-means and K-means++ lies in the initialization of cluster centroids. The standard K-means algorithm initializes centroids randomly, while K-means++ improves this initialization by selecting centroids in a more strategic and probabilistic manner. Here are the key distinctions:

**Centroid Initialization:**

* K-means: In the original K-means algorithm, centroids are initialized randomly from the available data points.
* K-means++: The K-means++ algorithm initializes the first centroid randomly, but subsequent centroids are chosen based on the distances to previously selected centroids. Data points farther away from existing centroids have a higher probability of being chosen.

**Impact on Clustering Results:**

* K-means: The random initialization in K-means can lead to different clusterings each time the algorithm is run. It is more sensitive to the initial placement of centroids, and there is a higher chance of converging to suboptimal solutions.
* K-means++: By strategically initializing centroids, K-means++ mitigates the sensitivity to initialization. It encourages a more spread-out initialization, which typically leads to better cluster formation and more reliable results. K-means++ tends to produce more stable and consistent clusters across different runs.

**Convergence and Efficiency:**

* K-means: The random initialization in K-means may require more iterations to converge, as the initial centroids might be far from their optimal positions. It can result in slower convergence and higher computational costs.
* K-means++: The improved initialization in K-means++ often leads to faster convergence since the initial centroids are better positioned. It reduces the number of iterations required to achieve convergence and can improve the algorithm's efficiency.
* Overall, K-means++ is a modification of the standard K-means algorithm that provides a more effective and robust initialization scheme. By selecting initial centroids strategically, it enhances the quality of the clustering results and reduces sensitivity to the initial random initialization. K-means++ is widely used in practice as it improves the performance and stability of the K-means clustering algorithm.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Load the Diamonds dataset
df = sns.load_dataset('diamonds')

# Select relevant features for clustering
features = ['carat', 'depth', 'price']

# Extract the selected features from the dataset
X = df[features]

# Perform K-means clustering with K-means++
k = 3  # Number of clusters
kmeans_pp = KMeans(n_clusters=k, init='k-means++', random_state=42)
kmeans_pp.fit(X)

# Perform standard K-means clustering
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(X)

# Add cluster labels to the dataset
df['cluster_pp'] = kmeans_pp.labels_
df['cluster'] = kmeans.labels_
# K_means++ is an improved version of K-means that helps to find better initial cluster centers,
#  which can lead to faster convergence and better clustering results compared to standard K-means.
df.head()
```
