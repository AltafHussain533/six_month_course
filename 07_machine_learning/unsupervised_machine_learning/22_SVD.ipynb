**Singular Value Decomposition (SVD)** is a powerful linear algebra matrix factorization method used to break down any real or complex matrix into three simpler, foundational constituent matrices. Unlike eigendecomposition—which can only be applied to square matrices—SVD can factorize any rectangular matrix of size m × n.
### **What is the SVD Definition?**
Mathematically, SVD states that any matrix A (of size m × n) can be factorized as:
$$A = U \Sigma V^T$$
Where the individual components represent specific structural properties:
- A(m × n): The original dataset or matrix.
- U(m × m): An orthogonal matrix containing the left-singular vectors. These vectors represent the row-space geometries and are the eigenvectors of the matrix A<sup>T</sup>A
- Σ(m × n): A diagonal matrix containing singular values sorted in descending order($\sigma_1 \ge \sigma_2 \ge \dots \ge 0$). These values quantify the strength, weight, or variance of each corresponding singular dimension.
- V<sup>T</sup>(n × n): The transpose of an orthogonal matrix containing right-singular vectors. These vectors represent the column-space geometries and are the eigenvectors of the matrix A<sup>T</sup>A.
### Why and How is SVD Used?
SVD acts like a structural prism for messy data. In engineering and data science, matrices represent complex operations or transformations. SVD breaks that transformation down into three clean, sequential geometric operations:
1. Rotation (V<sup>T</sup>): Aligns the data with the principal axes without warping it.
2. Stretching (Σ): Compresses or expands the data along those axes based on the scale of the singular values.
3. Rotation (U): Rotates the scaled data into its final destination space coordinates.By analyzing the singular values in Σ, an engineer can drop the smallest values to perform Truncated SVD, which eliminates noise and retains the core signal of the data using minimal storage. 
### Practical Applications of SVD1.
1. Image CompressionAn image is simply a matrix of pixel intensities. High-resolution images create massive matrices.
- How SVD helps: Running SVD on an image matrix yields singular values in Σ that drop off dramatically. By keeping only the top 20 or 50 largest singular values and discarding the rest, you compress the file size exponentially while keeping the image structurally recognizable.
2. Recommender Systems (The Netflix Prize)Streaming platforms utilize huge matrices where rows are users, columns are movies, and cell entries are user ratings. This matrix is mostly empty (sparse) because users only watch a fraction of the library.
- How SVD helps: SVD factorizes this sparse matrix into latent (hidden) feature spaces. It uncovers hidden dimensions like "User preference for Pakistani dramas" matched against "Content score for comedy," filling in the blanks to accurately predict what movie a user will enjoy next.
3. Principal Component Analysis (PCA)In the previous turns, we discussed PCA using eigenvalues of a covariance matrix.
- How SVD helps: In actual software implementation (like Python's scikit-learn), computing the covariance matrix explicitly can cause severe numerical precision errors on big datasets. Instead, PCA is under-the-hood calculated directly using SVD on the centered data matrix, offering cleaner numerical stability.
4. Natural Language Processing (Latent Semantic Analysis)In text processing, a "Document-Term Matrix" tracks how often specific words appear across thousands of articles.
- How SVD helps: SVD identifies underlying topics (concepts) by grouping synonymous words together. For instance, it recognizes that articles containing "Cricinfo," "Babar Azam," and "Gaddafi Stadium" all align along the same latent singular vector representing "Cricket in Pakistan."
5. Solving Ill-Conditioned Linear Systems (Pseudo-Inverse)When trying to solve equations of the form Ax = b, if matrix A is non-square or contains redundant data, traditional algebra fails because A cannot be inverted.
- How SVD helps: SVD is used to calculate the Moore-Penrose pseudo-inverse (A⁺). This allows engineers to find the closest, mathematically optimal "least-squares" solution to otherwise unsolvable engineering systems.

# **SVD (Singular value decomposition)**

A simple example of using SVD with sklearn in Python. The following example uses the UCI ML iris dataset (since the tips dataset isn't ideal for SVD). The sklearn.decomposition.TruncatedSVD class allows to perform linear dimensionality reduction.

```python
# import necessary libraries
from sklearn.decomposition import TruncatedSVD
from sklearn import datasets
import matplotlib.pyplot as plt

# load iris dataset as an example
iris = datasets.load_iris()
X = iris.data
y = iris.target

# SVD represent documents and terms in vectors 
svd = TruncatedSVD(n_components=2, algorithm='randomized', n_iter=100, random_state=122)
X_svd = svd.fit_transform(X)

# plot the result
plt.scatter(X_svd[:, 0], X_svd[:, 1], c=y, s=50, cmap='viridis')
plt.xlabel('Component 1')
plt.ylabel('Component 2')
plt.title('2-Component SVD')
plt.show()
```
